```python
# 5. Load ColQwen2-v1.0 Model
import torch
from colpali_engine.models import ColQwen2, ColQwen2Processor

MODEL_ID = 'vidore/colqwen2-v1.0'
device = 'cuda:0' if torch.cuda.is_available() else 'cpu'
dtype = torch.bfloat16 if torch.cuda.is_available() and torch.cuda.is_bf16_supported() else torch.float16

print(f'Loading {MODEL_ID} on {device} ({dtype})...', flush=True)
processor = ColQwen2Processor.from_pretrained(MODEL_ID)
model = ColQwen2.from_pretrained(MODEL_ID, torch_dtype=dtype, device_map=device).eval()
print('Model loaded successfully!', flush=True)
```

```python
# 6. Encode All 813 Document Images
from PIL import Image
import time

img_paths = [find_file(r['image_path']) for r in test_corpus]
corpus_ids = [r['id'] for r in test_corpus]
id_to_idx = {cid: i for i, cid in enumerate(corpus_ids)}

print(f'Encoding {len(img_paths)} document images on GPU...', flush=True)
t0 = time.time()
doc_embeddings = []
batch_size = 4

for b in range(0, len(img_paths), batch_size):
    b_paths = img_paths[b : b + batch_size]
    b_imgs = [Image.open(p).convert('RGB') if p.exists() else Image.new('RGB', (224, 224), (255, 255, 255)) for p in b_paths]
    b_proc = processor.process_images(b_imgs).to(device)
    with torch.no_grad():
        out = model(**b_proc)
        for i in range(len(b_imgs)):
            doc_embeddings.append(out[i].cpu())
    if (b // batch_size) % 25 == 0:
        print(f'  Encoded {min(b + batch_size, len(img_paths))}/{len(img_paths)} images...', flush=True)

print(f'Completed image encoding in {time.time() - t0:.1f}s!', flush=True)
```

```python
# 7. Late-Interaction MaxSim Retrieval Across All Query Suites
import numpy as np

def late_interaction_maxsim(query_emb: torch.Tensor, doc_emb: torch.Tensor) -> float:
    sim = torch.matmul(query_emb.float(), doc_emb.float().t())
    max_sim = torch.max(sim, dim=1).values
    return float(torch.sum(max_sim).item())

def score_query(q_text: str, tgt_id: str) -> float:
    b_q = processor.process_queries([q_text]).to(device)
    with torch.no_grad():
        q_emb = model(**b_q)[0].cpu()
    scores = np.array([late_interaction_maxsim(q_emb, d) for d in doc_embeddings])
    tgt_idx = id_to_idx[tgt_id]
    ranks = np.argsort(-scores)
    pos = np.where(ranks == tgt_idx)[0]
    if len(pos) == 0:
        return 0.0
    return 1.0 / (pos[0] + 1)

suites = [
    ('Primary Gen. (gpt-4o-mini, N=306)', primary_q, 0.4334, 0.2427, 0.1601),
    ('Secondary Gen. (Qwen2.5-7B, N=306)', qwen_q, 0.6292, 0.5802, 0.2685),
    ('Human-Authored Queries (N=100)', human_100_q, 0.6321, 0.5758, 0.5020),
    ('Human-Authored Queries (N=200)', human_200_q, 0.5766, 0.5549, 0.4909),
]

results_summary = []
for name, q_list, w_bm25, c_bm25, d_bge in suites:
    print(f'Scoring {name} ({len(q_list)} queries)...', flush=True)
    t_start = time.time()
    rrs = [score_query(q['query_text'], q['target_id']) for q in q_list]
    mrr = float(np.mean(rrs))
    # Bootstrap 95% CI
    rng = np.random.default_rng(42)
    boot = [float(np.mean(rng.choice(rrs, size=len(rrs), replace=True))) for _ in range(2000)]
    ci_l, ci_u = float(np.percentile(boot, 2.5)), float(np.percentile(boot, 97.5))
    results_summary.append({
        'Query Source': name,
        'N Probes': len(q_list),
        'Word BM25': w_bm25,
        'Char 3-5g BM25': c_bm25,
        'Dense BGE': d_bge,
        'ColQwen2 MRR': round(mrr, 4),
        '95% CI': f'[{ci_l:.4f}, {ci_u:.4f}]',
        'Elapsed (s)': round(time.time() - t_start, 1),
    })
    print(f'  -> MRR = {mrr:.4f} [{ci_l:.4f}, {ci_u:.4f}] in {time.time() - t_start:.1f}s', flush=True)
```
